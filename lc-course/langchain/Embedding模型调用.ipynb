Embedding Models嵌入模型的特点：将字符串作为输入，返回一个浮点数的列表（向量）
在NLP中，Embedding的作用就是将数据进行文本向量化

阿里千问模型访问方式：

```python
from langchain_community.embeddings import DashScopeEmbeddings

#初始化嵌入模型对象，其默认使用模型是：text-embedding-v1
embed = DashScopeEmbeddings()

#测试
print(embed.embed_query("hello")) #embed_query单条向量化
print(embed.embed_documents(['hello','world','messages']))#embed_documents批量向量化
```

本地Ollama模型访问方式：
通过langchain_ollama导入OllamaEmbeddings使用，其余不变。

```python
from langchain_ollama import OllamaEmbeddings
#初始化嵌入式模型对象，其默认使用模型是：text-embedding-v1
embed=OllamaEmbeddings(model="qwen3-embedding")
#测试
print(embed.embed_query("hello"))
print(embed.embed_documents(['hello','world','messages']))
```
